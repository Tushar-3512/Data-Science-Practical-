```python
import numpy as np

full_array = df.to_numpy()

print("Shape    :", full_array.shape)
print("Size     :", full_array.size)
print("Dimension:", full_array.ndim)
```

```python
numeric_df = df.select_dtypes(include='number')

print("Numeric columns found:", list(numeric_df.columns))
print("Numeric shape:", numeric_df.shape)
numeric_df.head()
```

```python
num_array = numeric_df.to_numpy()
print("Numeric array shape:", num_array.shape)
```

```python
half = num_array.shape[0] // 2
A = num_array[:half]
B = num_array[half:half*2]

print("A shape:", A.shape)
print("B shape:", B.shape)

addition       = A + B
subtraction    = A - B
multiplication = A * B

print("\nAddition:\n", addition[:3])
print("\nSubtraction:\n", subtraction[:3])
print("\nElement-wise Multiplication:\n", multiplication[:3])
```

```python
A_transpose = A.T
print("Original shape  :", A.shape)
print("Transposed shape:", A_transpose.shape)
A_transpose
```

```python
product = np.dot(A, B.T)   # (m,n) . (n,m) -> (m,m)
print("Product shape:", product.shape)
product[:3]
```

```python
n = A.shape[1]
square_matrix = A[:n, :n]

print("Square matrix:\n", square_matrix)
inverse_matrix = np.linalg.inv(square_matrix)
print("\nInverse:\n", inverse_matrix)
```
